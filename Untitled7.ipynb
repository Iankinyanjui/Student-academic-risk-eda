```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier, export_graphviz
import graphviz
```

```python
raw_data = {
    'Student_ID': [1001, 1002, 1003, 1004, 1005, 1001, 1006, 1007, 1008, 1009],
    'County': ['Nakuru', 'Narok', 'Nakuru', 'Narok', 'Kakamega',
               'Nakuru', 'Nairobi', None, 'Narok', 'Nakuru'],
    'Study_Hours_Per_Week': [15, 8, 20, None, 12, 15, 30, 10, 25, 18],
    'Attendance_Percent': [85, 60, 92, 55, 78, 85, 95, 65, 88, None],
    'Fee_Balance_Ksh': [0, 15000, 0, 25000, 5000, 0, 0, 18000, 2000, 12000],
    'Previous_Grade': [72, 48, 85, 45, 65, 72, 90, 50, 82, 58],
    'Student_Type': ['Regular', 'Regular', 'Scholarship', 'Regular',
                     'Scholarship', 'Regular', 'Regular', 'Scholarship',
                     'Regular', 'Regular'],
    'Academic_Risk': [0, 1, 0, 1, 0, 0, 0, 1, 0, 1]
}
```

```python
df = pd.DataFrame(raw_data)
print(df.shape)     # (10, 8)
print(df.dtypes)
print(df.describe())
```

```python
print(df.isnull().sum())      # County 1, Study_Hours 1, Attendance 1
print(df.duplicated().sum())  # 1 duplicate row (Student_ID 1001)
df = df.drop_duplicates()     # 10 rows becomes 9
```

```python
df['County'] = df['County'].fillna(df['County'].mode()[0])  # Nakuru
df['Study_Hours_Per_Week'] = df['Study_Hours_Per_Week'].fillna(df['Study_Hours_Per_Week'].median())  # 16.5
df['Attendance_Percent'] = df['Attendance_Percent'].fillna(df['Attendance_Percent'].median())        # 81.5
```

```python
sns.countplot(x='County', hue='Academic_Risk', data=df)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for ax, col in zip(axes.flat, ['Study_Hours_Per_Week','Attendance_Percent','Fee_Balance_Ksh','Previous_Grade']):
    sns.boxplot(x='Academic_Risk', y=col, data=df, ax=ax)
plt.tight_layout(); plt.show()
```

```python
df['Attendance_Category'] = pd.cut(df['Attendance_Percent'], bins=[0,70,85,100],
                                   labels=['Low','Medium','High'])
df['Has_Fee_Balance'] = (df['Fee_Balance_Ksh'] > 0).astype(int)
```

```python
enc = pd.get_dummies(df[['County','Student_Type']], drop_first=True, dtype=int)
num_cols = ['Study_Hours_Per_Week','Attendance_Percent','Fee_Balance_Ksh','Previous_Grade']
X = pd.concat([df[num_cols + ['Has_Fee_Balance']], enc], axis=1)
y = df['Academic_Risk']

X_scaled = X.copy()
X_scaled[num_cols] = StandardScaler().fit_transform(X[num_cols])
```

```python
corr_cols = num_cols + ['Has_Fee_Balance', 'Academic_Risk']
sns.heatmap(df[corr_cols].corr(), annot=True, cmap='coolwarm', center=0)
plt.show()
```

```python
dt = DecisionTreeClassifier(max_depth=3, random_state=42)
dt.fit(X_scaled, y)

dot = export_graphviz(dt, feature_names=X_scaled.columns,
                      class_names=['Not at risk','At risk'],
                      filled=True, rounded=True)
graphviz.Source(dot).render('tree', format='png', view=True)
```
